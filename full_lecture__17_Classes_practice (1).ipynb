# **instance** **method**= **methods bound to an instance**
Instance methods are functions defined inside a class that operate on an instance of the class that take self as their parameter , which refers to the specific instance calling the method. These methods can access and modify the attributes of the instance and call other instance methods.


```python
class House:
  #Constructor to initialize the house with a specific color
  def __init__(self,color):
    self.color=color #Assign the color to the house instance
    #instance method to display the color of the house
  def show_color(self):
    print(f"This house is painted {self.color}") #Access and print the house color
#create a house instance with 'blue' color
my_house=House("blue")
#call the instance method to show the color of the house
my_house.show_color()
```

# **Class methods: Bound to the class not to the object**

A class methodd is a method that is bound to the class itself , not to any one instance of the class . It can access or modify the **class**-**level** **state**. which is shared among all instances. Unlike instance methods that receive **self** , class methods receive **cls** as the first argument --refering to the class, not an individual object.
To define a class method , you use the @**classmethod** decorator above the method definition . These methods are commonly used for factory methods or when you want behavior that affects the class as a whole not just as a instance only.

```python
class House:
  def __init__(self,color,age):
    self.color=color
    self.age=age
  @classmethod
  def from_construction_year(cls, color, construction_year):
    #This class method creates a house objet by calculating its age
    return cls(color, 2024 - construction_year)
#create a new house using the class method as a factory
my_house=House.from_construction_year("white",2000)
#print the age of the house
print(my_house.age)
```

# **Static methods: Independent utility tools**
A Static method is a method that beloongs to a class bit doesn't access or modify the class or any of its instances. It doesn't take **self** (the instance) or **cls** (the class) as the first argument . Instead,static methods behave liike regular functions , but they're logically grouped with the class because they perform tasks relevant to that class's purpose .
Static methods are ideal for utility operations-- validations, calculations, or helpers-- that depent on the instance's or class's internal state . You define them using the @staticmethod decorator.

```python
class House:
  def __init__(self, color, area):
    self.color=color
    self.area=area #Area in square meters
  @staticmethod
  def is_valid_area(area):
    #Staticmethod to validate if the area value is reasonable
    return area >0
#use the static method to check if a given is valid before creating a house
print(House.is_valid_area(120))
print(House.is_valid_area(-50))
```
